### 🚀 Step 2: Run Live Cloud Extraction (`unitbytes/google-maps-leads-scraper`)
Grab your free Apify API token from **[Apify Console](https://apify.com/unitbytes/google-maps-leads-scraper?fpr=939u3w&fp_sid=gh_gmaps_leads_colab)** (includes `$5.00` free monthly usage credit) and run the cell below. `getpass` prompts for your token securely so it is never stored in notebook outputs:


```python
!pip install -q apify-client pandas

from getpass import getpass
import pandas as pd
from apify_client import ApifyClient

APIFY_TOKEN = getpass("Enter your Apify API Token (from https://apify.com/unitbytes/google-maps-leads-scraper?fpr=939u3w&fp_sid=gh_gmaps_leads_colab): ").strip()
if not APIFY_TOKEN:
    raise ValueError("Please enter a valid Apify API token to run live extraction.")

client = ApifyClient(APIFY_TOKEN)

run_input = {
    "searchQueries": ["dentists in Austin TX"],
    "maxPlacesPerQuery": 20,
    "reviewHookType": "most_relevant",
    "language": "en",
    "region": "us"
}

print("Running unitbytes/google-maps-leads-scraper...")
run = client.actor("unitbytes/google-maps-leads-scraper").call(run_input=run_input)
items = list(client.dataset(run["defaultDatasetId"]).iterate_items())

df_live = pd.DataFrame(items)
print(f"Extracted {len(df_live)} live records!")
df_live.to_csv("google-maps-leads-scraper-results.csv", index=False)
df_live.head(10)
```

### ⚡ Step 3: Call the Synchronous `<500ms` Google Maps B2B Leads REST API (RapidAPI Hub & Zyla API Hub)
Test the real-time synchronous `GET /v1/gmaps/search` endpoint on **[RapidAPI Hub](https://rapidapi.com/unitbytes/api/google-maps-b2b-leads-email-tech-stack-reviews-data-api)**:


```python
from getpass import getpass
import requests
import pandas as pd

RAPIDAPI_KEY = getpass("Enter your RapidAPI Key: ").strip()
resp = requests.get(
    "https://google-maps-b2b-leads-email-tech-stack-reviews-data-api.p.rapidapi.com/v1/gmaps/search",
    headers={"x-rapidapi-host": "google-maps-b2b-leads-email-tech-stack-reviews-data-api.p.rapidapi.com", "x-rapidapi-key": RAPIDAPI_KEY},
    params={"query": "dentists in Austin TX", "page_size": 3, "enrich_contacts": "true"},
    timeout=30,
)
resp.raise_for_status()
pd.DataFrame(resp.json().get("data", {}).get("items", [])).head()
```
